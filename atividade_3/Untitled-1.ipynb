```python
import pandas as pd
```

```python
path = r'H:\python\trabalho_3etapa\atividade_3\Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx' 

# Lê a planilha pura desde o começo para não perder nenhuma informação
dados = pd.read_excel(path, engine="openpyxl", header=None)
dados.head(15)
```

```python
# renomeia 
dados.rename(columns={dados.columns[0]: 'Regiao'}, inplace=True)

# Remove linhas vazias
dados = dados.dropna(subset=['Regiao'])
```

```python
# descobrir os nomes das colunas 

print(dados.columns.tolist())
```

```python
#  homens e mulheres por região com curso superior
tabela_superior = dados.iloc[:, [0, 1, 2]].copy()
tabela_superior.columns = ['Regiao', 'Homens', 'Mulheres']

print(" TABELA 1: curso superior total ")
print(tabela_superior.describe())
tabela_superior.to_excel(r"H:\python\trabalho_3etapa\atividade_3\curso_superior_total.xlsx", index=False)
```
