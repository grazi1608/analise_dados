```python
indicador_1 = pd.read_excel(path_horas, engine="xlrd", header=None)
indicador_1
```

```python
indicador_1 = indicador_1.iloc[8:41].copy()
```

```python
indicador_1.columns = [
    "uf_regiao",
    "total",
    "total_branca",
    "total_preta_parda",
    "homem_branca",
    "homem_preta_parda",
    "mulher_branca",
    "mulher_preta_parda",
]
indicador_1 = indicador_1.reset_index(drop=True)
```

```python
indicador_1
```

```python
regioes = ["Norte","Nordeste","Sudeste","Sul","Centro-Oeste"]

estados = indicador_1[~indicador_1["uf_regiao"].isin(["Brasil"] + regioes)]
```

```python
indicador_1
```

```python
estados = estados.rename(columns={"uf-regiao":"Estado"})
```

```python
estados
```

```python
estados = estados.rename(columns={"uf_regiao": "Estado"})
comp = comp.merge(estados, on="Estado", how="inner")
```

```python
comp
```

```python
ordem = comp.sort_values("mulheres_2026")["Estado"]
longo = comp.melt(
    id_vars=["Estado"],
    value_vars=["mulheres_2012", "mulheres_2026"],
    var_name="Ano",
    value_name="Participacao_mulheres",
)
longo["Ano"] = longo["Ano"].map({"mulheres_2012": "2012 T1", "mulheres_2026": "2026 T1"})

fig, ax = plt.subplots(figsize=(10, 10))
sns.barplot(
    data=longo,
    y="Estado",
    x="Participacao_mulheres",
    hue="Ano",
    order=ordem,
    ax=ax,
)
ax.axvline(50, color="gray", linestyle="--", linewidth=1) # MUDE A COR
ax.set_xlabel("Participação das mulheres entre as pessoas desocupadas (%)")
ax.set_ylabel("")
ax.set_title("Desocupação: participação feminina em 2012 T1 e 2026 T1")
ax.legend(title="aula 6")
fig.tight_layout()
plt.show()
```
