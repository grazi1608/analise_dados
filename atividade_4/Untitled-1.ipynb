```python
import pandas as pd
```

```python
path = r"H:\python\trabalho_3etapa\atividade_4\CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"
```

```python
# pip install openpyxl
# Lê a tabela pulando o topo 
dados = pd.read_excel(path, engine="openpyxl", header=2)
dados.head()



```

```python
# 5 - Crie uma tabela de população agregada por estado

dados_estado = dados.groupby("UF")[["População 2010 (Alterações de Limites até 2022)1", "População Censo 2022"]].sum().reset_index()
dados_estado.head()

```

```python
# 6 - Organize essa tabela por população que mais cresceu entre 2022 e 2010 (Faça a diferença se necessário)

dados_estado["Crescimento"] = dados_estado["População Censo 2022"] - dados_estado["População 2010 (Alterações de Limites até 2022)1"]

# maior para o menor
dados_estado_ordenado = dados_estado.sort_values(by="Crescimento", ascending=False)
dados_estado_ordenado.head()
```

```python
# 7 - Salve essa tabela nova em csv na pasta do seu projeto
dados_estado_ordenado.to_csv("populacao_por_estado.csv", sep=";", index=False)
```

```python
# 8 - Agora trabalhe por município

dados_municipio = dados.groupby(["UF", "NOME DO MUNICÍPIO"])[["População 2010 (Alterações de Limites até 2022)1", "População Censo 2022"]].sum().reset_index()
dados_municipio.head()
```

```python
# 9 - Organize essa tabela por população que mais cresceu entre 2022 e 2010 (Faça a diferença se necessário)

dados_municipio["Crescimento"] = dados_municipio["População Censo 2022"] - dados_municipio["População 2010 (Alterações de Limites até 2022)1"]
dados_municipio_ordenado = dados_municipio.sort_values(by="Crescimento", ascending=False)
dados_municipio.head()
```

```python
# 10 - Salve essa tabela nova em csv na pasta do seu projeto

dados_municipio_ordenado.to_csv("populacao_por_municipio.csv", sep=";", index=False)
```
